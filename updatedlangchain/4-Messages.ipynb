## Messages and their types

Messages are the Fundamental unit of context for models in Langchain.They represent the input and output of models,carrying both the content and metadata needed to represent the state of a conversation when interacting with an LLM. Messages are the objects that contain:

* Role - Identifies the message type(e.g.system,user)
* Content - Represent the actual content of the message(like text,image,audio,documents etc)
* Metadata - Optional fields such as response information,message IDs and token usage

Langchain Provide a standard message type that works across all model providers,ensuring consistent behavior reagardless of the model being called.


```python
import os 
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()
os.environ["Groq_API_Key"]= os.getenv("Groq_API_Key")
model = init_chat_model(
    model="openai/gpt-oss-20b",
    model_provider="groq",
)
```

```python
model.invoke("Please tell me a joke about programming.")
# This is human message given to the model. The output will be an AI message.
```

## 1. System message 
A SystemMessage gives instructions or establishes the behavior/personality of the AI.


```python
from langchain_core.messages import SystemMessage

message = SystemMessage(
    content="You are a helpful Python teacher."
)

print(message)

## With Model   
from langchain_core.messages import SystemMessage, HumanMessage

messages = [
    SystemMessage(content="You are a helpful Python teacher."),
    HumanMessage(content="What is a list in Python?")
]

response = model.invoke(messages)

print(response.content)
```

## 2. HumanMessage

HumanMessage represents the message/question sent by the user.

```python
from langchain_core.messages import HumanMessage

message = HumanMessage(
    content="What is Artificial Intelligence?"
)

print(message)

## With Model
response = model.invoke([
    HumanMessage(content="Explain Artificial Intelligence.")
])

print(response.content)
```

## 3. AIMessage

AIMessage represents a response generated by the AI model.

Normally, you don't manually create an AIMessage. LangChain creates it when the model responds.

```python
from langchain_core.messages import HumanMessage

messages = [
    HumanMessage(content="What is Python?")
]

response = model.invoke(messages)

print(response)
print(type(response))
```

## Text Prompt

Text Prompt are string - ideal for straightforward generation tasks where you don't need to retain conversation history.
Use text Prompt when:
*   You have a single,standalone request 
*   You don't need conversation history
*   You want minimal code complexity

```python
prompt = "What is Artificial Intelligence?"

response = model.invoke(prompt)

print(response.content)
```

## Message Prompt 

Alternatively,you can pass in a list of message to the model by providing a list of message objects.
A message prompt is a structured prompt used with chat-based LLMs where the input is represented as different message roles, such as system, human, AI, and tool. LangChain provides classes like ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate, and MessagesPlaceholder to create dynamic and structured conversational prompts.

SystemMessage → Instructions
HumanMessage  → User question
AIMessage     → Previous AI response (if any)


## 1. SystemMessagePromptTemplate

Used for system-level instructions.

```python
from langchain_core.prompts import SystemMessagePromptTemplate

system_prompt = SystemMessagePromptTemplate.from_template(
    "You are an expert {subject} teacher."
) 
```

## 2. HumanMessagePromptTemplate

Used for the user's question/input.

```python
from langchain_core.prompts import HumanMessagePromptTemplate

human_prompt = HumanMessagePromptTemplate.from_template(
    "Explain {topic} in simple language."
)
```

## 3. AIMessagePromptTemplate

Used when you want to include a previous AI response in the prompt.


```python
from langchain_core.prompts import AIMessagePromptTemplate

ai_prompt = AIMessagePromptTemplate.from_template(
    "Machine Learning is a branch of AI..."
)
```

## 4. MessagesPlaceholder

Used to insert a list of messages dynamically, especially conversation history.
This is very useful for chat memory.

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    MessagesPlaceholder("history"),
    ("human", "{question}")
])
history = [
    ("human", "My name is Rahul."),
    ("ai", "Nice to meet you, Rahul!")
]

messages = prompt.invoke({
    "history": history,
    "question": "What is my name?"
})

print(messages)
```

## 5. ChatPromptTemplate

Instead of manually creating every message, you can create a complete chat prompt:

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role}."),
    ("human", "Explain {topic} in simple language.")
])

messages = prompt.invoke({
    "role": "Python teacher",
    "topic": "decorators"
})

print(messages)
```
