## Model Integration with Google Gemini 



```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import SystemMessage, HumanMessage

load_dotenv()

llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0
)

messages = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="What is the capital of France?")
]

response = llm.invoke(messages)

print(response.content)
```

```python
print(response.content[0]["text"])
```

## Using Groq


```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq

# Load .env first
load_dotenv()

# Check API key
os.environ["Groq_API_Key"] = os.getenv("Groq_API_Key")

llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)

messages = "What is the capital of France?"

llm_response = llm.invoke(messages)

print(llm_response.content)
```

```python
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

model = init_chat_model(
    "openai/gpt-oss-20b",
    model_provider="groq",
    temperature=0
)

response = model.invoke("Why do parrots talk?")

print(response.content)
```

## Streaming

Most models can stream their output content while it is being generated.By displaying output progressively,streaming siginificantly improves user experience,particularly for longer responses.Calling stream() returns an itreator that yields output chunks as they are produces.You can use loop t process each chunk in real-time.


```python
content = """
Write a 200-word paragraph on Artificial Intelligence.
Write it as one continuous paragraph.
"""

response = model.invoke(content)

print(response.content)
```

```python
for chunk in model.stream("Why do parrots have the ability to mimic human speech?"):
    print(chunk, end=" ", flush=True)
```

## Batch 
Batching a collection of independent requests to a model can significantly improve performance and reduce cost,as the processing can be done in parallel:


```python
response = model.batch(["Why do parrots have the ability to mimic human speech?",
                        "What is the capital of France?",
                        "What is the weather like in New York City?"
                        ])
for r in response:
    print(r.content)
```

```python
response = model.batch(["Why do parrots have the ability to mimic human speech?",
                        "What is the capital of France?",
                        "What is the weather like in New York City?"],
                        config={
                         'max_concurrency': 5,
                     })
for r in response:
    print(r.content)
```
